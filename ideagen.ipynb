```python
!pip install -q groq
from google.colab import userdata
import os
from groq import Groq


os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")


client = Groq(api_key=os.environ.get("GROQ_API_KEY"))
MODEL = "openai/gpt-oss-120b"

SYSTEM_PROMPT = """
You are an AI College Project Idea Generator.

Your role is to help college students generate creative,
practical, innovative and technically feasible project ideas.

For every topic provided:
1. Generate exactly 3 different project ideas.
2. Give each idea a clear project title.
3. Briefly explain each idea.
4. Mention suitable technologies.
5. Prefer realistic ideas for college students.
6. Avoid repeating similar ideas.
"""


def generate_ideas(question, temperature=0.7):

    completion = client.chat.completions.create(

        model=MODEL,

        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": f"""
Generate 3 creative college project ideas based on:

{question}
"""
            }
        ],

        temperature=temperature,
        max_completion_tokens=500
    )

    response = completion.choices[0].message.content
    usage = completion.usage

    return response, usage



N = int(input("Enter number of questions: "))

for i in range(N):

    print("\n" + "=" * 70)
    print(f"QUESTION {i + 1}")
    print("=" * 70)

    question = input("Enter a college-related topic: ")

    response, usage = generate_ideas(
        question,
        temperature=0.7
    )

    print("\nPROMPT:")
    print(question)

    print("\nCOMPLETION:")
    print(response)

    print("\nTOKEN USAGE:")
    print("Prompt tokens    :", usage.prompt_tokens)
    print("Completion tokens:", usage.completion_tokens)
    print("Total tokens     :", usage.total_tokens)


choice = input(
    "\nDo you want to compare temperatures 0.0, 0.7 and 1.4? (yes/no): "
)

if choice.lower() == "yes":

    # Use the same topic that was entered in the main loop
    print("\n" + "=" * 70)
    print("TEMPERATURE COMPARISON")
    print("=" * 70)

    for temp in [0.0, 0.7, 1.4]:

        print("\n" + "-" * 70)
        print(f"TEMPERATURE = {temp}")
        print("-" * 70)

        response, usage = generate_ideas(
            question,
            temperature=temp
        )

        print("\nPROMPT:")
        print(question)

        print("\nCOMPLETION:")
        print(response)

        print("\nTOKEN USAGE:")
        print("Prompt tokens    :", usage.prompt_tokens)
        print("Completion tokens:", usage.completion_tokens)
        print("Total tokens     :", usage.total_tokens)
```
